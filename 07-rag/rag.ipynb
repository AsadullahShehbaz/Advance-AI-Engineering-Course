```python
# Load data using document loader

from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader("docs/profile.pdf")
documents = loader.load()

print(documents[0].page_content)
print(documents[0].metadata)
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(chunk_size=300,chunk_overlap=100,separators=["\n\n","\n"," ",""])

chunks = splitter.split_documents(documents)

print(len(documents))
print(len(chunks))
print(chunks[3].page_content)
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
```

```python
# Vector Store 

from langchain_chroma import Chroma

vector_store = Chroma.from_documents(
    documents=chunks,
    collection_name="langchain_rag",
    embedding=embeddings,
    persist_directory="./chroma_db",
    collection_metadata={"hnsw:space":"cosine"}
)
```

```python
retriever = vector_store.as_retriever(
    search_type="similarity",
    search_kwargs={"k":3}
)

documents = retriever.invoke("What is Asadullah AI Channel")

for doc in documents:
    print(doc.page_content)
```

```python
# Single Context String

context = "\n\n".join(document.page_content for document in documents)
print(context)
```

```python
from langchain_groq import ChatGroq
from dotenv import load_dotenv

load_dotenv()

llm = ChatGroq(
    model="openai/gpt-oss-120b"
)
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_template("""
You are a helpful AI assistant.

Answer the question using ONLY the provided context.

If the answer is not available in the context,
say that you do not have enough information.

Context:
{context}

Question:
{question}

Answer:
""")
```

```python
from langchain_core.output_parsers import StrOutputParser

chain = prompt | llm | StrOutputParser()
```

```python
question = "What is Asadullah AI Channel?"

response = chain.invoke({
    "context":context,
    "question":question
})

print(response)
```
