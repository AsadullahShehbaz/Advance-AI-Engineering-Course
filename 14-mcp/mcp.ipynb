```python
from langchain_groq import ChatGroq
from dotenv import load_dotenv
from langchain.messages import ToolMessage

load_dotenv()
model = ChatGroq(model="openai/gpt-oss-120b",  max_tokens=1024)
```

```python
config = {
    "mcpServers": {
        "langchain_docs": {
            "url": "https://docs.langchain.com/mcp"
        },
        "aws_docs": {
            "url": "https://knowledge-mcp.global.api.aws"
        },
        "microsoft_docs": {
            "url": "https://learn.microsoft.com/api/mcp"
        }
    }
}
```

```python
from langchain.agents import create_agent
from langchain.mcp import MCPAdapter


async def create_agent_with_tools(config,model):
    async with MCPAdapter(config) as adapter:
        tools = await adapter.list_tools()

        print("Available tools:",tools)
        for tool in tools:
            print(f"Tool name: {tool.name}")


        return create_agent(model=model, tools=tools)


agent = await create_agent_with_tools(config,model)

    
```

```python
response =  await agent.ainvoke({"messages": [{"role": "user", "content": "What is langchain?"}]})


for message in response["messages"]:

    if isinstance(message, ToolMessage):
        print(f"Tool Message: {message.content}")
        print(f"Status: {message.status}")
        print(f"Artifacts: {message.artifacts}")
    
print(response["messages"])
```
